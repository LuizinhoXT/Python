```python
# Intervalo
min = 0
max = 24
n = 8
```

```python
# Tamanho dos intervalos
dx = (max - min) / n
dx
```

- Obtendo a área pelo método da soma dos retângulos

```python
# transformar a funcao para ser utilizada em operações
x = smp.Symbol('x')
f = 2*x**2
f_x = smp.lambdify(x, f)
```

```python
# criar sequencia para eixo x
eixo_x = np.arange(min, max+dx, dx)
eixo_x
```

```python
# Definiçãp dos números para o eixo x
eixo_y = [f_x(i) for i in eixo_x]
```

```python
# Criação do gráfico
plt.bar(x = eixo_x, height = eixo_y, width= 2)
plt.plot(eixo_x, eixo_y, color='k')
plt.show()
```

```python
fx_dx = [i*dx for i in eixo_y]
sum(fx_dx)
```

- Obtendo a área pelo método da função integral

```python
x = smp.Symbol('x')
f = 2*x**2
f_int = smp.integrate(f)


f_x = smp.lambdify(x, f_int)

f_x(24) - f_x(0)
```

- Teste com um alto número para N

```python
# intervalo
minimo = 0
maximo = 24
n = 30000
# tamanho dos intervalos
dx = (maximo - minimo) / n
# transformar a funcao para ser utilizada em operações
x = smp.Symbol('x')
f = 2*x**2
f_x = smp.lambdify(x, f)
# criar a sequencia para o eixo x
eixo_x = np.arange(minimo, maximo+dx, dx)
# criar sequencia para eixo y
eixo_y = [f_x(i) for i in eixo_x]
# criar o gráfico
plt.bar(x = eixo_x, height = eixo_y, width=2)
plt.plot(eixo_x, eixo_y, color='k')
plt.show()
```

- Área aproximada:

```python
# intervalo
fx_dx = [i*dx for i in eixo_y]
sum(fx_dx)
```

- Resultados são parecidos quando aumentamos o valor de n

```python
# obter integral
x = smp.Symbol('x')
f = 2*x**2
funcao_integral = smp.integrate(f)
funcao_integral
# criar a função
f_x = smp.lambdify(x, funcao_integral)
# obter o valor da integral
f_x(24) - f_x(0)
```

- Calculando outra área abaixo da curva através do método da soma dos retângulos

```python
# intervalo
minimo = 50
maximo = -50
n = 10000
# tamanho dos intervalos
dx = (maximo - minimo) / n
# transformar a funcao para ser utilizada em operações
x = smp.Symbol('x')
f = 3*x**3 + x**2
f_x = smp.lambdify(x, f)
# criar a sequencia para o eixo x
eixo_x = np.arange(minimo, maximo+dx, dx)
# criar sequencia para eixo y
eixo_y = [f_x(i) for i in eixo_x]
# criar o gráfico
plt.bar(x = eixo_x, height = eixo_y, width=2)
plt.plot(eixo_x, eixo_y, color='k')
plt.show()
```

```python
fx_dx = [i*dx for i in eixo_y]
sum(fx_dx)
```

```python
# obter integral
x = smp.Symbol('x')
f = 3*x**3 + x**2
funcao_integral = smp.integrate(f)
funcao_integral
# criar a função
f_x = smp.lambdify(x, funcao_integral)
# obter o valor da integral
f_x(50) - f_x(-50)
```
